```python
import atoti as tt
import psycopg2
import pandas as pd
import os
from dotenv import load_dotenv
from pathlib import Path

load_dotenv(Path(".env"), override=True)

session = tt.Session.start()
print("Atoti session started!")
```

```python
print(dir(tt))
```

## Load Data Dari Supabase ke Atoti

```python
db_url = os.getenv("DATABASE_URL")

# Load materialized view sebagai tabel Atoti
df_harga = pd.read_sql("SELECT * FROM mv_harga_bulanan", 
                        psycopg2.connect(db_url))
df_indikator = pd.read_sql("SELECT * FROM mv_indikator_tahunan", 
                            psycopg2.connect(db_url))

print(f"mv_harga_bulanan: {len(df_harga)} baris")
print(f"mv_indikator_tahunan: {len(df_indikator)} baris")
print(df_harga.head())
```

## Buat DataMart & Cube

```python
# Buat table Atoti dari dataframe
tbl_harga = session.read_pandas(
    df_harga,
    table_name="Harga Pangan",
    keys=["year", "month", "country_name", "commodity_name"]
)

tbl_indikator = session.read_pandas(
    df_indikator,
    table_name="Indikator Ekonomi",
    keys=["year", "country_name"]
)

# Buat cube
cube = session.create_cube(tbl_harga, "Analisis Harga Pangan ASEAN")

print("Cube berhasil dibuat!")
print(cube.measures)
```

```python
cube.levels
```

## Custom Measures

```python
m = cube.measures
h = cube.hierarchies

# Measure custom
m["Rata-rata Harga"] = tt.agg.mean(tbl_harga["rata_harga"])
m["Harga Tertinggi"] = tt.agg.max(tbl_harga["max_harga"])
m["Harga Terendah"] = tt.agg.min(tbl_harga["min_harga"])
m["Total Observasi"] = tt.agg.sum(tbl_harga["total_observasi"])

print([k for k in m.keys()])
```

## Widget Visualisasi

```python
# Tampilkan URL dashboard Atoti
print(f"Atoti UI: {session.url}")

# Widget 1: Rata-rata harga per negara
widget1 = cube.query(
    m["Rata-rata Harga"],
    levels=[cube.levels["country_name"], cube.levels["year"]]
)

print(widget1)
```

## OLAP

### Rata-rata harga per negara per tahun

```python
q1 = cube.query(
    m["Rata-rata Harga"],
    m["Harga Tertinggi"],
    m["Harga Terendah"],
    levels=[cube.levels["country_name"], cube.levels["year"]]
)
print("Harga per Negara per Tahun")
q1
```

### Rata-rata harga per kategori per tahun

```python
q2 = cube.query(
    m["Rata-rata Harga"],
    m["Total Observasi"],
    levels=[cube.levels["category"], cube.levels["year"]]
)
print("\nHarga per Kategori")
q2
```

### Rata-rata harga per bulan

```python
q3 = cube.query(
    m["Rata-rata Harga"],
    levels=[cube.levels["country_name"], cube.levels["month"]]
)
print("\nHarga per Bulan")
q3
```

### Top 10 Komoditas Termahal per Tahun

```python
q4 = cube.query(
    m['Rata-rata Harga'], m['Total Observasi'],
    levels=[cube.levels['commodity_name'], cube.levels['year']]
)
q4_reset = q4.reset_index()
top10 = (
    q4_reset.groupby('year', group_keys=False)
    .apply(lambda x: x.nlargest(10, 'Rata-rata Harga'))
    .reset_index(drop=True)
)

top10[['year','commodity_name','Rata-rata Harga']]
```

### Analisis Komoditas per Negara

```python
q5 = cube.query(
    m['Rata-rata Harga'], m['Harga Tertinggi'], m['Harga Terendah'],
    levels=[cube.levels['country_name'], cube.levels['category']]
)
q5
```

## Visualisasi hasil OLAP

### Rata rata harga per tahun

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
q1_reset = q1.reset_index()
fig, ax = plt.subplots(figsize=(8, 5))

for country, group in q1_reset.groupby("country_name"):
    ax.plot(group["year"], group["Rata-rata Harga"], marker='o', label=country)

ax.set_title("Rata-rata Harga Pangan per Negara per Tahun")
ax.set_xlabel("Tahun")
ax.set_ylabel("Harga")
ax.legend()
plt.tight_layout()
plt.savefig("viz_harga_negara.png", dpi=150)
plt.show()
```

### Rata rata harga per kategori

```python
q2_reset = q2.reset_index()
q2_pivot = q2_reset.pivot(index="category", columns="year", values="Rata-rata Harga")

fig, ax = plt.subplots(figsize=(10, 5))
q2_pivot.plot(kind="bar", ax=ax)
ax.set_title("Rata-rata Harga per Kategori Komoditas")
ax.set_xlabel("Kategori")
ax.set_ylabel("Harga")
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.savefig("viz_harga_kategori.png", dpi=150)
plt.show()
```

### Rata rata harga per bulan

```python
q3_reset = q3.reset_index()

fig, ax = plt.subplots(figsize=(10, 5))
for country, group in q3_reset.groupby("country_name"):
    ax.plot(group["month"], group["Rata-rata Harga"], marker='o', label=country)

ax.set_title("Tren Rata-rata Harga per Bulan")
ax.set_xlabel("Bulan")
ax.set_ylabel("Harga")
ax.set_xticks(range(1, 13))
ax.legend()
plt.tight_layout()
plt.savefig("viz_harga_bulan.png", dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
colors = plt.cm.Blues_r(np.linspace(0.2, 0.8, 10))
for i, year in enumerate([2024, 2025]):
    data = top10[top10['year'] == year].nlargest(10, 'Rata-rata Harga')
    axes[i].barh(data['commodity_name'], data['Rata-rata Harga'], color=colors)
    axes[i].set_title(f'Top 10 Komoditas Termahal ({year})', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('Rata-rata Harga')
    axes[i].invert_yaxis()
    for bar, val in zip(axes[i].patches, data['Rata-rata Harga']):
        axes[i].text(bar.get_width()*0.01, bar.get_y()+bar.get_height()/2,
                     f'{val:,.0f}', va='center', fontsize=7)
plt.tight_layout()
plt.savefig('viz_top10_komoditas.png', dpi=150)
plt.show()
```

```python
q5_reset = q5.reset_index()

pivot = q5_reset.pivot(
    index='country_name',
    columns='category',
    values='Rata-rata Harga'
)

plt.figure(figsize=(10,6))

sns.heatmap(
    pivot,
    annot=True,
    fmt='.0f',
    cmap='YlOrRd'
)

plt.title('Heatmap Rata-rata Harga per Negara dan Kategori')
plt.show()
```
